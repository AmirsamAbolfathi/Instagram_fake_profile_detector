```python
import pandas as pd
import pickle as pk
from sklearn.model_selection import train_test_split
from sklearn import preprocessing,metrics,ensemble
```

```python
data = pd.read_csv("Instagram_fake_profile_dataset.csv")
data
```

```python
data = data.drop_duplicates()
data.duplicated().sum(), data.isnull().sum()
```

```python
data.shape
```

```python
x = data.drop("fake",axis=1)
y = data["fake"]
x_train, x_test, y_train, y_test = train_test_split(x, y, random_state=42, train_size=0.8)
```

```python
scalar = preprocessing.StandardScaler()
x_train = pd.DataFrame((scalar.fit_transform(x_train)))
x_test = scalar.transform(x_test)

x_train.describe().T
```

```python
model_forest = ensemble.RandomForestClassifier()
model_forest.fit(x_train,y_train)
y_predict_forest = model_forest.predict(x_test)
metrics.accuracy_score(y_test,y_predict_forest)*100
```

```python
with open("scalar.pkl","wb") as file:
    try:
        pk.dump(scalar, file)
        print("scalar ----> successful")
    except:
        print("unsuccessful")

with open("model_forest.pkl","wb") as file:
    try:
        pk.dump(model_forest,file)
        print("model_forest ----> successful")
    except:
        print("unsuccessful")
```
